```python
# Import 10x Visium data
adata2 = sc.read_10x_h5("/mnt/datadisk/Jordan/Data/Visium/MpBC2/filtered_feature_bc_matrix.h5")

# Number of duplicates BEFORE
print(adata2.var_names.duplicated().sum())

# Make var_names unique
adata2.var_names_make_unique()

# Check if var_names are unique
print(adata2.var_names.duplicated().sum())  # Should be 0

# Add metadata to obs
sample = "MpBC2"
adata2.obs['sample'] = sample  # Example metadata, replace with actual sample name
adata2.obs['batch'] = 'batch1'  # Example batch, replace with actual batch name

# Load metadata annotation from CSV file
metadata_df2 = pd.read_csv("/mnt/datadisk/Jordan/Data/Visium/MpBC2/Annotations_old_with_unannotated.csv")

# Ensure Barcode is string type
metadata_df2['Barcode'] = metadata_df2['Barcode'].astype(str)  

# Ensure that the index corresponds to cell names
metadata_df2.set_index('Barcode', inplace=True)

# Display first few rows of metadata
metadata_df2.head()  

# Merge metadata with adata.obs
adata2.obs = adata2.obs.join(metadata_df2, how='left')

print(adata2.obs['Annotations_old_with_unannotated'].isna().sum())

# Include in barcode names the sample name
adata2.obs_names = [f"{sample}_{barcode}" for barcode in adata2.obs_names]

# Display the first few rows of the updated obs
print(adata2.obs.head())
```

```python
# Loop through additional samples and concatenate
samples = ["MpBC1", "MpBC2", "MpBC3", "MpBC4", "MpBC5", "MpBC6", "MpBC7", "MpBC8", 
          "MpBC9", "MpBC10", "MpBC11", "MpBC13", "MpBC14", "MpBC15", "MpBC16"]

for sample in samples:
    adata_sample = sc.read_10x_h5(f"/mnt/datadisk/Jordan/Data/Visium/{sample}/filtered_feature_bc_matrix.h5")
    
    # Make var_names unique
    adata_sample.var_names_make_unique()
    
    # Add metadata to obs
    adata_sample.obs['sample'] = sample  # Example metadata, replace with actual sample name
    adata_sample.obs['batch'] = 'batch1'  # Example batch, replace with actual batch name
    
    # Load metadata annotation from CSV file
    metadata_df_sample = pd.read_csv(f"/mnt/datadisk/Jordan/Data/Visium/{sample}/Annotations_old_with_unlabels.csv")
    
    # Ensure Barcode is string type
    metadata_df_sample['Barcode'] = metadata_df_sample['Barcode'].astype(str)  
    
    # Ensure that the index corresponds to cell names
    metadata_df_sample.set_index('Barcode', inplace=True)
    
    # Merge metadata with adata.obs
    adata_sample.obs = adata_sample.obs.join(metadata_df_sample, how='left')
    
    print(adata_sample.obs['Annotations_old'].isna().sum(), "cells without annotations in", sample)
    
    # Include in barcode names the sample name
    adata_sample.obs_names = [f"{sample}_{barcode}" for barcode in adata_sample.obs_names]
    
    # Concatenate with the main AnnData object
    if sample == "MpBC1":
        print("Skipping MpBC1 concatenation as it is the first loop.")
        adata_concat = adata_sample
        continue
    else:
        print(f"Concatenating sample: {sample}")
        adata_concat = sc.concat(
            [adata_concat, adata_sample],
            join='outer',
            label='sample',
            keys=[adata_concat.obs['sample'].unique()[0], sample]
)
```

```python
print(adata_concat)
print(adata_concat.obs['sample'].value_counts())

print("adata shape:", adata_concat.shape)  # Shape of the AnnData object (cells x genes)
print("Number of cells:", adata_concat.n_obs)  # obs: cell information (e.g., cell barcodes, metadata)
print("Number of genes:", adata_concat.n_vars)  # var: gene information (e.g., gene symbols)
print("Cell barcodes:", adata_concat.obs_names[:5])  # Display first 5 cell barcodes
print("Gene names:", adata_concat.var_names[:5])  # Display first 5 gene
print("Cell metadata (obs):", adata_concat.obs.head())  # Display first few rows of cell metadata
print("Gene metadata (var):", adata_concat.var.head())  # Display first few rows of gene metadata
print("Unstructured data (uns):", adata_concat.uns.keys())  # Display keys in unstructured data
print("Reduced embeddings (obsm):", adata_concat.obsm.keys())  # Display keys in reduced embeddings
print("Gene embeddings (varm):", adata_concat.varm.keys())  # Display keys in gene embeddings
print("Layers:", adata_concat.layers.keys())  # Display keys in layers
print("Raw data:", adata_concat.raw)  # Display raw data if available
```

```python
cell_type_list = []
cell_type = adata_concat.obs['Annotations_old'].unique()
for cell in cell_type:
    if pd.isna(cell):
        continue
    cell_type_list.append(cell)
print("Unique cell types:", cell_type_list)
print("Number of unique cell types:", len(cell_type_list))
```

```python
path = "/mnt/datadisk/Jordan/Data/Visium/CellType_matrix/"

adata_subsets = {}

for cell_type in cell_type_list:
    print(f"Processing cell type: {cell_type}")
    adata_subsets[cell_type] = adata_concat[adata_concat.obs['Annotations_old'] == cell_type].copy()
    print(f"Number of cells in {cell_type}: {adata_subsets[cell_type].n_obs}")

    # Save each subset to a file
    adata_subsets[cell_type].write(f"{path}{cell_type.replace(' ', '_')}.h5ad")
    print(f"Saved {cell_type.replace(' ', '_')}.h5ad")
    # adata.write_loom(f"{path}{cell_type.replace(' ', '_')}.loom")
    # print(f"Saved {cell_type.replace(' ', '_')}.loom")
```
